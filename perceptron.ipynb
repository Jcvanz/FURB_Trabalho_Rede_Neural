## Jogador ou Piloto?

O perceptron foi treinado para dizer se um atleta é jogador de futebol (classe 0) ou piloto de F1 (classe 1), usando três características: Ciência, Física e Filosofia.

---

### Dados usados no treinamento

| Atleta     | Ciência | Física | Filosofia | Classe |
|------------|---------|--------|-----------|--------|
| Neymar     |    1    |   0    |     0     |   0    |
| Messi      |    1    |   0    |     1     |   0    |
| Barichello |    1    |   1    |     0     |   1    |
| Massa      |    1    |   1    |     1     |   1    |

- **Classe 0** -> Jogador de futebol  
- **Classe 1** -> Piloto de F1

---
## 1. Bibliotecas

Foram usadas apenas o `numpy` para os cálculos e o `matplotlib` para o gráfico de erros.

```python
import numpy as np
import matplotlib.pyplot as plt

# fixando um seed pra os resultados não mudarem a cada execução
np.random.seed(42)
```

---
## 2. Dados de Treinamento

Os dados foram montados de acordo com a tabela da aula.

```python
atletas = ['Neymar', 'Messi', 'Barichello', 'Massa']

# cada linha é um atleta: [Ciência, Física, Filosofia]
X = np.array([
    [1, 0, 0],   # Neymar
    [1, 0, 1],   # Messi
    [1, 1, 0],   # Barichello
    [1, 1, 1],   # Massa
])

# 0 = jogador, 1 = piloto
y = np.array([0, 0, 1, 1])

print('Tabela de treinamento:')
print(f'{"Atleta":<12} {"Ciência":<10} {"Física":<10} {"Filosofia":<12} {"Classe"}')
print('-' * 55)
for i, nome in enumerate(atletas):
    print(f'{nome:<12} {X[i][0]:<10} {X[i][1]:<10} {X[i][2]:<12} {y[i]}')
```

---
## 3. Perceptron

O perceptron funciona assim:

- Multiplica cada entrada pelo seu peso e soma tudo (mais o bias)
- Se o resultado for maior ou igual a zero, a saída e 1; se nao, e 0
- Se errou, os pesos são ajustados: `w = w + eta * erro * x`
- Repete isso até acertar tudo ou chegar no limite de épocas

```python
class Perceptron:
    def __init__(self, n_entradas, taxa_aprendizado=0.1, epocas=100):
        self.taxa_aprendizado = taxa_aprendizado
        self.epocas = epocas

        # pesos e bias começam com valores aleatórios pequenos
        self.pesos = np.random.uniform(-0.5, 0.5, n_entradas)
        self.bias  = np.random.uniform(-0.5, 0.5)

        # guarda quantos erros tiveram em cada época
        self.historico_erros = []

    def ativacao(self, u):
        # função degrau: retorna 1 ou 0
        return 1 if u >= 0 else 0

    def prever(self, x):
        u = np.dot(self.pesos, x) + self.bias
        return self.ativacao(u)

    def treinar(self, X, y):
        print('Início do treinamento')
        print(f'Pesos iniciais : {np.round(self.pesos, 4)}')
        print(f'Bias inicial   : {self.bias:.4f}')
        print()

        for epoca in range(1, self.epocas + 1):
            erros = 0

            for xi, yi in zip(X, y):
                saida = self.prever(xi)
                erro  = yi - saida

                if erro != 0:
                    # ajuste dos pesos
                    self.pesos += self.taxa_aprendizado * erro * xi
                    self.bias  += self.taxa_aprendizado * erro
                    erros += 1

            self.historico_erros.append(erros)
            print(f'Epoca {epoca:>3} | Erros: {erros} | Pesos: {np.round(self.pesos, 4)} | Bias: {self.bias:.4f}')

            # parou de errar, entao o treinamento acabou
            if erros == 0:
                print(f'\nTreinamento finalizado na época {epoca} sem erros')
                break
        else:
            print('\nLimite de épocas atingido')

        print(f'\nPesos finais: {np.round(self.pesos, 4)}')
        print(f'Bias final: {self.bias:.4f}')
```

---
## 4. Treinamento

O perceptron foi rodado com taxa de aprendizado de **0.1** e até **100 épocas**

```python
perceptron = Perceptron(n_entradas=3, taxa_aprendizado=0.1, epocas=100)
perceptron.treinar(X, y)
```

---
## 5. Resultado

Comparação entre o que o perceptron classificou e o que era esperado para cada atleta

```python
print('Classificação final:')
print(f'{"Atleta":<12} {"Esperado":<12} {"Obtido":<12} {"Certo?"}')
print('-' * 48)

acertos = 0
for nome, xi, yi in zip(atletas, X, y):
    pred = perceptron.prever(xi)
    esp  = 'Jogador' if yi   == 0 else 'Piloto'
    obt  = 'Jogador' if pred == 0 else 'Piloto'
    ok   = 'Sim' if pred == yi else 'Não'
    if pred == yi:
        acertos += 1
    print(f'{nome:<12} {esp:<12} {obt:<12} {ok}')

taxa = acertos / len(y) * 100
print(f'\nAcertos: {acertos} de {len(y)} ({taxa:.0f}%)')
```
